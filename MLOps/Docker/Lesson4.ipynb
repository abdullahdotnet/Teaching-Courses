### **1. What is Docker?**

#### **Definition**
Docker is an open-source platform designed to help developers build, ship, and run applications in isolated environments called **containers**.

#### **Why Use Docker?**
- **Portability**: Applications run consistently across different environments (e.g., local machines, servers, cloud).
- **Isolation**: Each container has its own resources (e.g., CPU, memory, network), preventing interference between applications.
- **Efficiency**: Containers are lightweight compared to traditional virtual machines (VMs) because they share the host OS kernel.
- **Speed**: Containers start quickly and use fewer resources than VMs.

#### **Analogy**
Think of containers as **shipping containers** for software. They package everything an application needs to run (code, libraries, dependencies) into a single unit.

### **2. Key Docker Concepts**

| **Term**            | **Explanation**                                                                                  |
|----------------------|--------------------------------------------------------------------------------------------------|
| **Image**            | A lightweight, standalone, and immutable package containing the code and dependencies of an app. |
| **Container**        | A running instance of an image. It is isolated from other containers and the host system.         |
| **Dockerfile**       | A text file containing instructions to build a Docker image.                                      |
| **Docker Hub**       | A cloud-based repository for storing and sharing Docker images.                                   |
| **Volume**           | A way to persist data generated by containers.                                                   |
| **Orchestration**    | Managing multiple containers using tools like Docker Compose or Kubernetes.                      |

### **3. Installing Docker**

#### **Prerequisites**
- A modern OS: Windows, macOS, or Linux.

#### **Installation Steps**
1. **Windows/Mac:**
   - Download Docker Desktop: [https://www.docker.com/products/docker-desktop](https://www.docker.com/products/docker-desktop).
   - Install and run Docker Desktop.
   - Ensure WSL2 (Windows Subsystem for Linux 2) is enabled on Windows.

2. **Linux:**
   - Install Docker Engine:
     ```bash
     sudo apt update
     sudo apt install docker.io
     ```
   - Start Docker:
     ```bash
     sudo systemctl start docker
     sudo systemctl enable docker
     ```
   - Verify the installation:
     ```bash
     docker --version
     ```

### **4. Basic Docker Commands**

#### **Checking Docker Installation**
```bash
docker --version
docker info
```

#### **Pulling an Image**
Download a pre-built image from Docker Hub.
```bash
docker pull nginx
```

#### **Running a Container**
Run an image as a container.
```bash
docker run -d -p 8080:80 nginx
```
- `-d`: Detached mode (runs in the background).
- `-p`: Maps host port (`8080`) to container port (`80`).

#### **Listing Containers**
```bash
docker ps         # Shows running containers
docker ps -a      # Shows all containers (including stopped ones)
```

#### **Stopping a Container**
```bash
docker stop <container_id>
```

#### **Removing a Container**
```bash
docker rm <container_id>
```

#### **Removing an Image**
```bash
docker rmi <image_name>
```

### **5. Building and Running a Simple Dockerized Application**

#### **Step 1: Create a Simple Python Application**
Create a file named `app.py`:
```python
from flask import Flask

app = Flask(__name__)

@app.route('/')
def home():
    return "Hello, Docker!"

if __name__ == '__main__':
    app.run(host='0.0.0.0', port=5000)
```

#### **Step 2: Create a Dockerfile**
Create a file named `Dockerfile` in the same directory:
```dockerfile
# Use the official Python image
FROM python:3.9-slim

# Set the working directory
WORKDIR /app

# Copy the current directory contents into the container
COPY . .

# Install dependencies
RUN pip install flask

# Expose port 5000
EXPOSE 5000

# Define the command to run the application
CMD ["python", "app.py"]
```

#### **Step 3: Build the Docker Image**
Run the following command in the terminal:
```bash
docker build -t flask-app .
```
- `-t`: Tags the image (name it `flask-app`).

#### **Step 4: Run the Docker Container**
Run the container from the built image:
```bash
docker run -d -p 5000:5000 flask-app
```

#### **Step 5: Access the Application**
Visit `http://localhost:5000` in your browser. You should see "Hello, Docker!".

### **6. Persisting Data with Volumes**

#### Why Use Volumes?
Data inside a container is temporary. Volumes allow data to persist even if the container is deleted.

#### Example: Bind Mount
```bash
docker run -v $(pwd)/data:/app/data flask-app
```
This command mounts the `data` directory on the host to `/app/data` inside the container.

### **7. Docker Compose (Optional)**
Docker Compose simplifies managing multi-container applications.

#### Example `docker-compose.yml`:
```yaml
version: '3'
services:
  web:
    build: .
    ports:
      - "5000:5000"
```

Run the application with:
```bash
docker-compose up
```

- Containers are lightweight, fast, and consistent across environments.
