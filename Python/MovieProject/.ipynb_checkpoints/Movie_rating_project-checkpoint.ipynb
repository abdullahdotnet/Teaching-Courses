```python
movielist = []
with open('movieratings.txt','r') as file1:
    # lines = len(file1.readlines())
    # for i in range(lines):
    #     data = file1.readline()
    #     movielist.append(data)
    data = file1.readlines()
    for item in data:
        keyvalue = item.split(':')
        key = keyvalue[0]
        value = str(keyvalue[1])
        value = value.rstrip('\n')
        value = value.rstrip(',')
        value = value.lstrip(' ')
        value = value.lstrip('[')
        value = value.rstrip(']')
        value = value.strip(',')
        movies[key] = value

print(movies)
```

```python

    


def show_all_movies():
    print("========================")
    for key,value in movies.items():
        print(key,'====>',value)
    print("========================")
def add_rating(movie_title, rating):
    if movie_title not in movies.keys():
      movies[movie_title] = []
      movies[movie_title].append(rating)
    else:
      movies[movie_title].append(rating)

def get_average_rating(movie_title):
    ratings = movies[movie_title]
    movieaverage = sum(ratings)/len(ratings)
    print("===========================")
    print('Average Rating',movieaverage)
    print("===========================")

def top_rated_movie():
    max = 0
    topratedmovie = ''
    for key,value in movies.items():
      ratings = movies[key]
      movieaverage = sum(ratings)/len(ratings)
      if movieaverage > max:
        max = movieaverage
        topratedmovie = key
    print("===============================")
    print("Top rated movie", topratedmovie)
    print("===============================")
```

```python
flag = True
while flag == True:
    choice = int(input("1: Add Movie Rating\n2: Get Average Movie Rating:\n3: Top Rated Movie\n4: Show All Movies\n5: Exit Menu\n"))

    if choice == 1:
        movie = input('Enter movie title: ')
        rating = float(input('Enter movie rating: '))
        add_rating(movie,rating)

    elif choice == 2:
        movie = input('Enter movie title: ')
        get_average_rating(movie)

    elif choice == 3:
        top_rated_movie()
    elif choice == 4:
        show_all_movies()
    elif choice == 5:
        flag = False
```
