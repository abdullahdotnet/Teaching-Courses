## loc[ ] vs. iloc[ ] in Pandas:

Both loc[ ] and iloc[ ] are used to select data from a Pandas DataFrame or Series, but they differ in how they access the data.

1. loc[ ]: Label-based Indexing
loc[ ] is used to select data by the label of rows and columns. It allows you to access groups of rows and columns by their names (or labels).
It works with index labels (both row and column labels) and can also accept boolean arrays for filtering.

```python
student_dict = {
    'name':['Abdullah','John','Jacob','Emma','Sarah','Shamraiz'],
    'iq':[100,90,120,80,0,0],
    'marks':[80,70,100,50,0,0],
    'package':[10,7,14,2,0,0]
}
```

```python
data = pd.DataFrame(student_dict)
```

```python
data.head()
```

```python
data.loc[0:3,['name','marks']]
```

changing index from 0 1 2 3 to names to clear how loc used labels not position

```python
data.set_index('name',inplace=True)
data
```

```python
data.loc[['Abdullah':'John'],['name','marks']]
```

2. iloc[ ]: Integer-based Indexing
iloc[ ] is used to select data by integer position. It allows you to access rows and columns using their numerical positions, like an array index in Python.
It works purely with integer positions, similar to how Python lists or NumPy arrays work.

```python
data.iloc[0:3,[0,2]]
```

```python
data.iloc[0]
```

```python
normal_data = {
    "size": ["small", "medium", "large"],
    "normal_price": [5.4, 8.7, 11.4]
}
normal = pd.DataFrame(normal_data)

premium_data = {
    "size": ["large", "medium", "small"],
    "premium_price": [11.0, 11.7, 6.9]
}
premium = pd.DataFrame(premium_data)
```

```python
# The merge() function is used to join two DataFrames (normal and premium) based on a common column.
# In this case, we are merging the 'normal' DataFrame with the 'premium' DataFrame.
# right=premium: Specifies that the 'premium' DataFrame is the right DataFrame in the merge.
# left_on='size': Uses the 'size' column from the 'normal' DataFrame as the key for the join.
# right_on='size': Uses the 'size' column from the 'premium' DataFrame as the key for the join.

# The result is a new DataFrame (prices) that contains the 'size' column, 
# along with both 'normal_price' and 'premium_price' columns for matching sizes.

prices = normal.merge(right=premium,left_on='size',right_on='size')
```

```python
prices.head()
```

```python
prices['diff'] = abs(prices['normal_price'] - prices['premium_price'])
```
