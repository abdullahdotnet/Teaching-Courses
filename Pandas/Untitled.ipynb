```python
import pandas as pd
import numpy as np
from scipy.stats import zscore

np.random.seed(42)
data_dict = {f'Var{i+1}': np.random.randint(1, 101, 10) for i in range(25)}

data = pd.DataFrame(data_dict)

# print("Sample DataFrame:\n", data)

z_scores = data.apply(zscore)
print(z_scores)

data['Index'] = z_scores.mean(axis=1)

# print("\nDataFrame with Z-Score Index:\n", data)
```
